# 03 — Embeddings

Compare **HuggingFace**, **OpenAI**, **Ollama**, and **Cohere** embedding providers.

| Provider | Type | Model | Dim |
|---|---|---|---|
| HuggingFace | Local | `BAAI/bge-small-en-v1.5` | 384 |
| OpenAI | Cloud | `text-embedding-3-small` | 1536 |
| Ollama | Local | `nomic-embed-text` | 768 |
| Cohere | Cloud | `embed-english-light-v3.0` | 384 |

**Config source:** `configs/default.yaml` → `embeddings:`
**Secrets source:** `.env` (`OPENAI_API_KEY`, `COHERE_API_KEY`, `HUGGINGFACEHUB_API_TOKEN`)

```python
import sys
import os
sys.path.append("..")

from rag_pipeline.utils import load_notebook_config, get_secret
from rag_pipeline.embeddings import build_embeddings

cfg, REPO = load_notebook_config()
print("Repo root:       ", REPO)
print("Default provider:", cfg.embeddings.get("provider"))
print("Default model:   ", cfg.embeddings.get("model"))
```

**HuggingFace**

```python
hf = build_embeddings(dict(cfg.embeddings))
print(f"HuggingFace dim: {len(hf.embed_query('What is a neural network?'))}")
```

**OpenAI**

```python
openai_emb = build_embeddings({
    "provider": "openai",
    "model": "text-embedding-3-small",
})
print(f"OpenAI dim: {len(openai_emb.embed_query('test'))}")
```

**Ollama**

```python
ollama_emb = build_embeddings({
    "provider": "ollama",
    "model": "nomic-embed-text",
})
print(f"Ollama dim: {len(ollama_emb.embed_query('test'))}")
```

**Cohere**

```python
cohere_emb = build_embeddings({
    "provider": "cohere",
    "model": "embed-english-light-v3.0",
})
print(f"Cohere dim: {len(cohere_emb.embed_query('test'))}")
```

```python
PROVIDERS = {
    "huggingface": {"provider": "huggingface", "model": cfg.embeddings.get("model", "BAAI/bge-small-en-v1.5")},
    "openai":      {"provider": "openai",      "model": "text-embedding-3-small"},
    "ollama":      {"provider": "ollama",      "model": "nomic-embed-text"},
    "cohere":      {"provider": "cohere",      "model": "embed-english-light-v3.0"},
}

print(f"{'Provider':<15} {'Model':<35} {'Dim':>6}")
print("-" * 58)
for name, spec in PROVIDERS.items():
    try:
        emb = build_embeddings(spec)
        dim = len(emb.embed_query("test"))
        print(f"{name:<15} {spec['model']:<35} {dim:>6}")
    except Exception as e:
        print(f"{name:<15} {spec['model']:<35} {'—':>6}  ({type(e).__name__})")
```
